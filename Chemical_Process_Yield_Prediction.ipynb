# Chemical Process Yield Prediction Using Multiple Linear Regression

**Model:** Multiple Linear Regression  
**Dataset:** Simulated chemical-process operating data

> **Note:** The dataset used in this project is simulated for demonstration and is not experimental plant data.

## 1. Project Objective

The aim of this project is to predict chemical process yield (%) from selected operating conditions:

- Temperature (°C)
- Pressure (bar)
- Flow Rate (L/min)
- Concentration (%)
- Residence Time (min)

Multiple Linear Regression is used to model the relationship between these variables and process yield.

## 2. Import Libraries

The project uses NumPy, Pandas, Matplotlib, and Scikit-learn.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

## 3. Dataset

A simulated dataset is created to represent different chemical process operating conditions. The data is used only to demonstrate the machine learning workflow and should not be treated as experimental data.

```python
np.random.seed(42)

n = 300

temperature = np.random.uniform(60, 120, n)
pressure = np.random.uniform(1, 5, n)
flow_rate = np.random.uniform(2, 10, n)
concentration = np.random.uniform(10, 40, n)
residence_time = np.random.uniform(5, 30, n)

noise = np.random.normal(0, 2.5, n)

yield_percent = (
    35
    + 0.35 * temperature
    + 2.5 * pressure
    - 0.8 * flow_rate
    + 0.55 * concentration
    + 0.45 * residence_time
    + noise
)

yield_percent = np.clip(yield_percent, 0, 100)

df = pd.DataFrame({
    "Temperature_C": temperature,
    "Pressure_bar": pressure,
    "Flow_Rate_L_min": flow_rate,
    "Concentration_percent": concentration,
    "Residence_Time_min": residence_time,
    "Yield_percent": yield_percent
})

df.head()
```

## 4. Data Exploration

```python
print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns)

print("\nData types:")
print(df.dtypes)

print("\nStatistical summary:")
display(df.describe())
```

## 5. Data Cleaning

Missing values and duplicate rows are checked before building the model.

```python
print("Missing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())
```

## 6. Visualization

The plots are used to see how the process variables vary with yield.

```python
features = [
    "Temperature_C",
    "Pressure_bar",
    "Flow_Rate_L_min",
    "Concentration_percent",
    "Residence_Time_min"
]

for feature in features:
    plt.figure(figsize=(6, 4))
    plt.scatter(df[feature], df["Yield_percent"], alpha=0.7)
    plt.xlabel(feature)
    plt.ylabel("Yield (%)")
    plt.title(f"{feature} vs Yield")
    plt.grid(True)
    plt.show()
```

## 7. Input Variables and Target

`X` contains the process operating conditions, while `y` contains the yield to be predicted.

```python
X = df[features]
y = df["Yield_percent"]

print("Input shape:", X.shape)
print("Target shape:", y.shape)
```

## 8. Train-Test Split

The data is divided into training and testing sets. The model is trained on the training set and evaluated on unseen test data.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42
)

print("Training inputs:", X_train.shape)
print("Testing inputs:", X_test.shape)
```

## 9. Model Training

Multiple Linear Regression is fitted using the training data.

```python
model = LinearRegression()
model.fit(X_train, y_train)

print("Model training completed.")
```

## 10. Predictions

The trained model is used to predict yield for the test set.

```python
y_pred = model.predict(X_test)

results = pd.DataFrame({
    "Actual_Yield": y_test.values,
    "Predicted_Yield": y_pred
})

display(results.head(10))
```

## 11. Model Evaluation

The model is evaluated using MAE, RMSE, and R².

```python
mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print(f"MAE: {mae:.2f} percentage points")
print(f"RMSE: {rmse:.2f} percentage points")
print(f"R² Score: {r2:.3f}")
```

## 12. Actual vs Predicted Yield

This plot compares the observed yield values with the values predicted by the model.

```python
plt.figure(figsize=(6, 5))
plt.scatter(y_test, y_pred, alpha=0.7)

minimum = min(y_test.min(), y_pred.min())
maximum = max(y_test.max(), y_pred.max())

plt.plot([minimum, maximum], [minimum, maximum], linestyle="--")
plt.xlabel("Actual Yield (%)")
plt.ylabel("Predicted Yield (%)")
plt.title("Actual vs Predicted Yield")
plt.grid(True)
plt.show()
```

## 13. Coefficient Interpretation

The regression coefficients show the estimated change in yield for a one-unit change in each input variable, while the other variables are held constant.

```python
coefficient_table = pd.DataFrame({
    "Variable": features,
    "Coefficient": model.coef_
})

display(coefficient_table)

print("Intercept:", model.intercept_)
```

### Chemical Engineering Interpretation

- **Temperature:** Within the simulated range, a positive coefficient indicates an association between higher temperature and higher predicted yield.
- **Pressure:** A positive coefficient indicates higher predicted yield as pressure increases, within the modeled range.
- **Flow Rate:** A negative coefficient indicates lower predicted yield as flow rate increases, within the modeled range.
- **Concentration:** A positive coefficient indicates higher predicted yield at higher concentration, within the modeled range.
- **Residence Time:** A positive coefficient indicates higher predicted yield with longer residence time, within the modeled range.

These relationships are based on the simulated dataset and do not establish causation.

## 14. Sample Prediction

A new set of operating conditions is passed to the trained model to estimate process yield.

```python
new_conditions = pd.DataFrame({
    "Temperature_C": [95],
    "Pressure_bar": [3],
    "Flow_Rate_L_min": [5],
    "Concentration_percent": [25],
    "Residence_Time_min": [18]
})

predicted_yield = model.predict(new_conditions)

print(f"Predicted yield: {predicted_yield[0]:.2f}%")
```

## 15. Conclusion

The model demonstrates how Multiple Linear Regression can be used to estimate chemical process yield from operating conditions. The results show the relationship captured by the selected variables and provide a simple example of data-driven process analysis.

Because the dataset is simulated, the model should not be treated as a validated industrial process model.
